```python
# Question 10: Write a code that generates prime numbers up to n using a generator.
```

```python
def prime_generator(n):
    for num in range(2, n + 1):
        is_prime = True
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                is_prime = False
                break
        if is_prime:
            yield num

for value in prime_generator(30):
    print(value)
```

```python
# Question 42: How does Python achieve polymorphism through method overriding?
```

```python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

class CardPayment(Payment):
    def pay(self, amount):
        print(f"Paid {amount} using card")

class UpiPayment(Payment):
    def pay(self, amount):
        print(f"Paid {amount} using UPI")

payments = [CardPayment(), UpiPayment()]

for payment in payments:
    payment.pay(1000)
```

```python
# Question 50: Develop a Person class with private attributes name and email, and methods to set and get the email.
```

```python
class BankAccount:
    def __init__(self, balance=0):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Invalid withdrawal")

    def check_balance(self):
        return self.__balance

account = BankAccount(1000)
account.deposit(500)
account.withdraw(300)
print(account.check_balance())
```

```python
# Question 54: Create two decorators, and apply them to a single function. Ensure that they execute in the order they are applied.
```

```python
def message_decorator(message):
    def decorator(func):
        def wrapper(*args, **kwargs):
            print(f"{func.__name__}: {message}")
            return func(*args, **kwargs)
        return wrapper
    return decorator

@message_decorator("Function is starting")
def greet(name):
    print(f"Hello, {name}")

greet("Aman")
```

```python
# Question 67: What is the purpose of a getter method in Python? Provide an example demonstrating the use of a getter method using property decorators.
```

```python
class CountUp:
    def __init__(self, limit):
        self.current = 1
        self.limit = limit

    def __iter__(self):
        return self

    def __next__(self):
        if self.current <= self.limit:
            value = self.current
            self.current += 1
            return value
        raise StopIteration

for number in CountUp(5):
    print(number)
```

```python
# Question 69: What is the purpose of the @property decorator in Python? Provide an example illustrating its usage.
```

```python
class Person:
    def __init__(self, name):
        self._name = name

    @property
    def name(self):
        return self._name

    @name.setter
    def name(self, new_name):
        self._name = new_name

person = Person("Aman")
person.name = "Riya"
print(person.name)
```
